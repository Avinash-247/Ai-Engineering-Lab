```python
from langchain_chroma  import Chroma

vectors_store=Chroma(
    collection_name="firstivs",
    embedding_function=embedding,
    persist_directory="./vectors_store/exp_vectors"
)
```

```python
from langchain_core.documents import Document
doc=Document(
    page_content="Acceptable Use Policy Acceptable Use Policy Company devices and networks are intended primarily for business use.",
    metadata={"source": "./text_Doc/acceptable_use_policy.txt"}
)

Document(
    page_content="Anti Harassment Policy The organization maintains a workplace free from harassment, bullying, and discrimination. Complaints will be investigated promptly and handled",
    metadata={"source": "./text_Doc/anti_harassment_policy.txt"}
)

```

```python
vectors_store.add_documents(documents=[doc])
```

```python
r=vectors_store.as_retriever().invoke("Acceptable")
```

```python
for resu in r:
    print(resu)
```
