```python
from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_messages([
    ("system", "you are an expert in{topic}."),
    ("user", "your openion on {topic} and {audience} in one line. ")
])

chain= prompt|llm
responsess=chain.invoke({"topic":"Humans",
                       "audience":"intelligent_Sysyems"}
                       )
print(responsess.content)
```

```python
##  output Structure

# whe we get responsefrom llm we want outputs in a specific structure 

from langchain_core.output_parsers import StrOutputParser

output=StrOutputParser()

chain=prompt |llm | output
res=chain.invoke({"topic":"sports",
                  "audience":"students"})
print(res)
```

```python
from pydantic import BaseModel, Field
from langchain_core.prompts import ChatPromptTemplate
class  SupportTicket(BaseModel):
    issues_category :str = Field(description="one of :  billling")
    resolved :bool= Field(description="Wether the issue is resolved are not")
    next_action:str = Field(descriptio="what should i suppose to do now ")

prompt=ChatPromptTemplate.from_messages([
    ("system","yud summerize customer support"),
    ("user","My last delivery  is not arrived"),
    ("assistant","How would i help you")
])

chain= prompt |llm.with_structured_output(SupportTicket, method='Json_schema')
respose=chain.invoke({})
respose
```
