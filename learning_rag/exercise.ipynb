```python
from langchain_community.document_loaders import TextLoader
# here first i want to do with all text files, due to the folling the guidence, 
# i am going with one text file, with all file ill do it later After complteof this
text_loader=TextLoader(
    file_path="./exercise/India.txt",
    encoding="utf-8",

)

document=text_loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter=RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=0,
    separators=["\n"]
)
```

```python
chunk=splitter.split_documents(document)
```

```python
len(chunk)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embedging=GoogleGenerativeAIEmbeddings(model="text-embedding-005")

result=embedging.embed_query("what is HR")
```

```python
len(result)
```

```python
from langchain_chroma import Chroma

storage = Chroma.from_documents(
    documents=chunk,
    embedding=embedging,
    collection_name="country",
    persist_directory="./exercise_store/vector_store"
)
```

```python
vector_sto=storage.add_documents(documents=chunk)
```

```python
print(type(vector_sto))
```

```python
print(type(storage))
```

```python
# as_retriver andai only chrom lo unte net vastadi list lo unte radhu

result=vector_sto.as_retriever().invoke("what are the best places visit in india")
```

```python
#chunck storage andai chroma lo undhi ka batti retriver vachadi
retriver=storage.as_retriever()
```

```python
result=retriver.invoke("what are the 5 top places in india")
```

```python
for results in result:
    print(results)
```

```python
for doc in result:
    print(doc.page_content)
```
